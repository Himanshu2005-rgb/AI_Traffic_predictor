```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
df = pd.read_csv("traffic_congestion_data.csv")

print("First 5 rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
print(df.info())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check numerical columns
numeric_columns = [
    "hour",
    "temperature",
    "rainfall",
    "vehicle_count",
    "average_speed",
    "accident"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Remove rows containing missing values
df = df.dropna()

print("\nCleaned Dataset Shape:")
print(df.shape)
```

```python
# Convert day into numerical values
day_mapping = {
    "Monday": 0,
    "Tuesday": 1,
    "Wednesday": 2,
    "Thursday": 3,
    "Friday": 4,
    "Saturday": 5,
    "Sunday": 6
}

df["day_number"] = df["day_of_week"].map(day_mapping)

# Create peak-hour feature
df["peak_hour"] = df["hour"].apply(
    lambda x: 1 if x in [7, 8, 9, 17, 18, 19, 20] else 0
)

# Create traffic density feature
df["vehicles_per_speed"] = (
    df["vehicle_count"] / df["average_speed"]
)

print("\nFeature Engineered Data:")
print(df.head())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="congestion_level"
)

plt.title("Traffic Congestion Distribution")
plt.xlabel("Congestion Level")
plt.ylabel("Number of Records")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x="hour",
    y="vehicle_count",
    marker="o"
)

plt.title("Vehicle Count by Hour")
plt.xlabel("Hour")
plt.ylabel("Vehicle Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="vehicle_count",
    y="average_speed",
    hue="congestion_level",
    s=100
)

plt.title("Vehicle Count vs Average Speed")
plt.xlabel("Vehicle Count")
plt.ylabel("Average Speed")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="congestion_level",
    y="vehicle_count"
)

plt.title("Vehicle Count by Congestion Level")

plt.show()
```

```python
features = [
    "hour",
    "day_number",
    "temperature",
    "rainfall",
    "vehicle_count",
    "average_speed",
    "accident",
    "peak_hour",
    "vehicles_per_speed"
]

X = df[features]

y = df["congestion_level"]

# Encode target
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\nTarget Classes:")
print(label_encoder.classes_)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    max_depth=10
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\nModel Accuracy:")
print(f"{accuracy * 100:.2f}%")

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=label_encoder.classes_
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title("Traffic Congestion Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
importance = pd.DataFrame({
    "Feature": features,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nFeature Importance:")
print(importance)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=importance,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance")

plt.show()
```

```python
import os

# Create models folder if it doesn't exist
os.makedirs("models", exist_ok=True)

# Save trained model
joblib.dump(
    model,
    "models/traffic_model.pkl"
)

# Save label encoder
joblib.dump(
    label_encoder,
    "models/label_encoder.pkl"
)

print("Model saved successfully!")
```
